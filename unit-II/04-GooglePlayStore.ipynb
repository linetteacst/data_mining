**Pregunta B:** Escriban la instrucción exacta usando `dropna() `pasándole el parámetro `subset=['Android Ver', 'Type']` para eliminar únicamente las 3 filas que tienen esos nulos. Verifiquen que el total de registros en el DataFrame apenas disminuyó.

```python
df = df.dropna(subset=['Android Ver', 'Type'])
```

```python
df.isnull().sum()
```

**Pregunta C**: Rellenar los 1,500 nulos de Rating con el promedio global es un error, ya que las apps de la categoría FAMILY (Familia) no se califican igual que las de DATING (Citas). Utilizando la lógica de la semana pasada, escriban un bloque combinando .groupby('Category') y .transform() para rellenar los NaN de la columna Rating usando la **mediana de su propia categoría**.

```python
df['Rating'] = df['Rating'].fillna(df.groupby('Category')['Rating'].transform('median'))
df.isnull().sum()


```

**Pregunta D**: Revisen la columna `Size` (Tamaño de la app). Notarán que no tiene nulos matemáticos (`NaN`), pero muchas filas dicen `"Varies with device"`. Para un modelo matemático, ese texto es basura indigerible. Escriban una instrucción usando `df['Size'].replace(...)` para convertir exactamente esa frase de texto en un nulo matemático real usando `np.nan` de `NumPy`. ¿Por qué es mejor convertirlo a `NaN` en esta etapa en lugar de ponerle un tamaño falso como `"0 MB"`?

```python
df['Size'] = df['Size'].replace('Varies with device', np.nan)
```

**R** = para que sea más fácil identificar si el valor realmente no existe en vez de asignar un valor que podría suceder de forma natural dentro del dataset.
