```python
import random

random_list = [random.random() for _ in range(5_000_000)]

print("Random list length:", len(random_list))
```

```python
def standard_deviation(list):
    n = len(list)
    
    aux=0
    for num in list:
        aux += num

    mean = aux / n

    aux = 0
    for num in list:
        aux += (num - mean)**2
        
    return (aux / n)**0.5
```

```python
print(standard_deviation(random_list))
```

```python
import numpy as np

arr = np.array(random_list)
print("Standard deviation using NumPy:", np.std(arr))
```

```python
sumf = 0
summ = 0
for s, f in zip(df['Sex'], df['Fare']):
    if s == 'female':
        sumf+= f

    else:
        summ+= f


print("average for woman:", sumf / len(df[df['Sex'] == 'female']))
print("average for male:", summ / len(df[df['Sex'] == 'male']))
```

```python
q1 = df['Fare'].quantile(0.25)
q3 = df['Fare'].quantile(0.75)


iqr = q3 - q1

minimum = q1 - 1.5*iqr
maximum = q3 + 1.5*iqr

print("Minimum:", minimum)
print("Maximum:", maximum)


auxm = auxma =0

for fare in df['Fare']:
    if fare < minimum:
        auxm += 1
    elif fare > maximum:
        auxma += 1

print("Number of outliers below minimum:", auxm)
print("Number of outliers above maximum:", auxma)
print("Total: ", auxm+auxma)
```
