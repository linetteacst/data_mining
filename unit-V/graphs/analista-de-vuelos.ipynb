```python
import pandas as pd
import networkx as nx
```

```python
df = pd.read_csv('vuelos_10k.csv')

df.head()
ruta = ["Origen", "Destino"]

# quitar nulos
print("Registros sin origen o destino:", df[ruta].isna().any(axis=1).sum())
df = df.dropna(subset=ruta).copy()

# To upper
for columna in ruta:
    df[columna] = df[columna].str.upper()

# quitar self loops
df = df[df['Origen'] != df['Destino']]

# valores menores a 0
df = df[df['Distancia_km'] >= 0]

# drop duplicates
df = df.drop_duplicates(subset=ruta, keep='first')
print('Df final: ', len(df))
```

```python
G = nx.from_pandas_edgelist(df, 'Origen', 'Destino', edge_attr=['Distancia_km'])

print('Edges: ',G.number_of_edges())
print('Nodes: ',G.number_of_nodes())
```

```python
matrix = nx.adjacency_matrix(G).todense()

mb_matriz = matrix.nbytes / (1024*1024)
mb_lista = df.memory_usage(deep=True).sum() / (1024*1024)

print('List: ',mb_lista)
print("Matriz: ",mb_matriz)

print(mb_matriz/mb_lista)
```

## ¿Cuántas veces más pesada es la Matriz en comparación con la Lista de Aristas y por qué sucede esto?

La matriz es 64 veces mas grande, debido a que la información que se guarda dentro de la matriz resulta ser en muchos casos redundante, o sea, se guarda la misma información en distintos registros pues sigue un size de NxN.
