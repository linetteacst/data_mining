**Instrucciones**: Investiguen y apliquen en un Jupyter Notebook nuevo las siguientes tres técnicas de sumarización sobre el dataset del Titanic. Redacten en una celda Markdown qué significa el resultado que obtuvieron:

```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
df.describe()
```

**Asimetría (Skewness)**: Investiguen cómo aplicar la función .skew() en la columna Age. Si el resultado es positivo o negativo, ¿qué significa físicamente sobre la edad de los pasajeros en el barco? (¿Había más jóvenes o más ancianos?).

```python
df['Age'].skew()
```

    **R** = el resultado es positivo, lo que significa que la edad de los pasajeros se encuentra en los valores "jóvenes", se concentraban más en la izquierda

**Curtosis (Kurtosis)**: Investiguen cómo aplicar la función .kurt() en la columna Fare. ¿Qué significa este número sobre el "grosor" de las colas de la distribución? ¿Confirma esto la existencia de los outliers que encontramos hoy en clase?

```python
df['Fare'].kurt()
```

    **R** = la kurtosis se refiere a que tan afectada se ve la distribución por valores outliers. Refleja que tan pesadas son las colas comparado a una distribución normal. En este caso, como se obtuvo un valor muy grande se confirma la existencia de los outliers encontrados dentro de la columna Fare.

**Tablas Cruzadas (Cross-Tabulation)**: Investiguen la función pd.crosstab(). Generen una tabla cruzada entre la clase del pasajero (Pclass) y si sobrevivió (Survived). Esta técnica de sumarización matricial es el equivalente a una Tabla Dinámica en Excel. ¿Qué patrón logran visualizar rápidamente al cruzar ambas variables?

```python
pd.crosstab(df['Pclass'], df['Survived'])
```

    **R** = mientras mayor era la clase, mayor era la posibilidad de sobrevivir para los pasajeros de dicha clase. Primera clase > segunda clase > tercera clase.
